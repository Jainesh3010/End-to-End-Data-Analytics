```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df=pd.read_csv("zepto_sales_dataset.csv")
df
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.shape
```

```python
df.drop_duplicates()
```

```python
df["Cost"] = df["Current Price"] * 0.7 * df["Orders"]
df["Profit"]=df["Total Revenue"]-df["Cost"]
df["Total_discount"]=df["Discount"]*df["Orders"]
df["Gross_Revenue"] = df["Total Revenue"]+ df["Total_discount"]

```

```python
df.head()
```

```python
print("Total Cost Is:",df["Cost"].sum().astype(int))
```

```python
print("Total Profit Is:",df["Profit"].sum().astype(int))
```

```python
print("Total Revenue After Discount Is:",df["Total Revenue"].sum().astype(int))
```

```python
print("Total Discount Is:",df["Total_discount"].sum().astype(int))
```

```python
print("Total Gross Revenue With Discount Is:",df["Gross_Revenue"].sum().astype(int))
```

```python
df.columns=df.columns.str.strip()
```

```python
df.columns=df.columns.str.replace(" ","_")
```

```python
df
```

```python
df["Date"] = np.random.choice(
    pd.date_range("2024-01-01","2024-12-31"),
    size=len(df)
)
```

```python
df
```

```python
df.to_csv("ZeptoData.csv")
```

```python
df[["Cost","Profit"]].corr(numeric_only=True)
```

```python
df["Month"]=pd.to_datetime(df["Date"]).dt.month
```

```python
df.groupby(["Product_Name","Category","City"])[["Total_Revenue","Cost","Profit","Total_discount","Gross_Revenue"]].sum().reset_index()
```

```python
df.groupby("Product_Name")["Profit"].sum().plot(kind="bar")
plt.xlabel("Products")
plt.ylabel("Profit")
plt.show()
```

```python
df.groupby("City")[["Total_Revenue","Cost","Profit","Total_discount","Gross_Revenue"]].sum().reset_index(0)
```

```python
df.groupby("City")[["Total_Revenue","Cost","Profit","Total_discount","Gross_Revenue"]].sum().plot(kind="bar")
plt.ticklabel_format(style="plain", axis="y")
plt.xlabel("City")
plt.ylabel("Number")
plt.show()
```

```python
df.groupby("City")["Profit"].sum().plot(kind="bar")
plt.xlabel("City")
plt.ylabel("Profit")
plt.show()
```
